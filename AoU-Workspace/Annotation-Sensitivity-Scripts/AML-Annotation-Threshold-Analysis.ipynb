# AML ANNOTATION THRESHOLD ANALYSIS 
This script submits SAIGE-GENE+ rare variant set-based gene association analyses for four different annotation groups. On the smallest chromosome, chromosome 21, this was performed as part of a sensitivity analysis to determine the optimal threshold for each category to be used in the main analysis. Specifically, this included determining the threshold for LOFTEE to classify loss-of-function variants, the threshold for SpliceAI scores to identify splicing variants, the threshold for AlphaMissense scores to classify missense variants, and the use of synonymous variants as a negative control.

```python
# Imports the 'os' module, which allows interaction with the operating system (e.g., environment variables, file paths).
import os 

# Retrieves the value of the environment variable 'WORKSPACE_BUCKET'.
my_bucket = os.getenv('WORKSPACE_BUCKET')

# Displays the value of 'my_bucket' (i.e., the workspace bucket URL or path).
my_bucket
```

# COMMAND 

```python
%%writefile lof_sensitivity_analysis.sh
#!/bin/bash

# Exit immediately if a command exits with a non-zero status
set -o errexit

# Treat unset variables as an error (prevents using undefined variables)
set -o nounset

# Loop through the populations
for pop in EUR; do
    echo "Processing population: ${pop}"
    
    for pheno in aml_phenotype; do 
    echo "Processing pop: ${pop} | phenotype: ${pheno} | chr: ${chrom}"
    
    step2_SPAtests.R \
        --bedFile="${plink_bed}" \
        --bimFile="${plink_bim}" \
        --famFile="${plink_fam}" \
        --SAIGEOutputFile="${output}/${pop}_${pheno}_lof_chr${chrom}" \
        --LOCO=FALSE \
        --AlleleOrder=alt-first \
        --minMAF=0 \
        --minMAC=0.5 \
        --GMMATmodelFile="${null_glmm}/${pop}_${pheno}_null_glmm.rda" \
        --varianceRatioFile="${null_glmm}/${pop}_${pheno}_null_glmm.varianceRatio.txt" \
        --sparseGRMFile="${grm}/${pop}_GRM_relatednessCutoff_0.05_5000_randomMarkersUsed.sparseGRM.mtx" \
        --sparseGRMSampleIDFile="${grm}/${pop}_GRM_relatednessCutoff_0.05_5000_randomMarkersUsed.sparseGRM.mtx.sampleIDs.txt" \
        --groupFile="${vat}/${pop}_annotation_test/lof_annotation_chr${chrom}.txt" \
        --annotation_in_groupTest=LOFTEE-HC,LOFTEE-HC:LOFTEE-LC,LOFTEE-HC:LOFTEE-LC:HIGH-IMPACT \
        --maxMAF_in_groupTest=0.0001,0.001,0.01 \
        --is_output_markerList_in_groupTest=TRUE \
        --is_fastTest=TRUE
        
    done
done
```

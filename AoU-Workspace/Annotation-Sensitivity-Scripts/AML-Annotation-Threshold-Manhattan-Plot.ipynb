```r
library(data.table)
library(ggplot2)
library(tidyverse)
```

```r
lof <- fread("sensitivity_analysis/EUR_aml_phenotype_lof_chr21")
splice <- fread("sensitivity_analysis/EUR_aml_phenotype_splice_chr21")
missense <- fread("sensitivity_analysis/EUR_aml_phenotype_missense_chr21")
synonymous <- fread("sensitivity_analysis/EUR_aml_phenotype_synonymous_chr21")
```

```r
options(repr.plot.width = 20,  # width in inches
        repr.plot.height = 10) # height in inches
```

```r
lof %>% 
  filter(Group != "Cauchy") %>% 
  mutate(
    neg_log_pvalue = -log10(Pvalue),
    max_MAF = factor(max_MAF)   # <- convert to factor
  ) %>% 
  ggplot(aes(x = Region, y = neg_log_pvalue, color = Group, shape = max_MAF)) +
  geom_jitter(size = 5, alpha = 0.5, width = 10) +
  theme_bw() +
  theme(
    legend.text = element_text(size = 14),
    legend.title = element_text(size = 18), 
    axis.marker
  ) + 
  paletteer::scale_colour_paletteer_d("LaCroixColoR::Lemon") 
```

```r
lof %>% 
    filter(Group != "Cauchy") %>% 
    mutate(neg_log_pvalue = -log10(Pvalue)) %>% 
ggplot() +
geom_jitter(aes(x = Region, 
               y = neg_log_pvalue, 
               color = Group,
              
               size = 1, 
               alpha = 0.5), 
               width = 5) + 
theme_bw() + 
theme(legend.text = element_text(size = 14),
      legend.title = element_text(size = 18))
```

```r
splice %>% 
    filter(Group != "Cauchy") %>% 
    mutate(neg_log_pvalue = -log10(Pvalue)) %>% 
ggplot() +
geom_jitter(aes(x = Region, 
               y = neg_log_pvalue, 
               color = Group, 
               size = neg_log_pvalue, 
               alpha = 0.5)) + 
theme_bw() + 
theme(legend.text = element_text(size = 14),
      legend.title = element_text(size = 18))
```

```r
synonymous %>% 
    filter(Group != "Cauchy") %>% 
    mutate(neg_log_pvalue = -log10(Pvalue)) %>% 
ggplot() +
geom_jitter(aes(x = Region, 
               y = neg_log_pvalue, 
               color = Group, 
               size = neg_log_pvalue, 
               alpha = 0.5)) + 
theme_minimal() + 
theme(legend.text = element_text(size = 14), 
      legend.title = element_text(size = 18))
```

```r
synonymous %>% 
    filter(Group != "Cauchy") %>% 
    mutate(neg_log_pvalue = -log10(Pvalue)) %>% 
ggplot(aes(x = neg_log_pvalue, fill = Group, group = Group)) +
  geom_density(alpha = 0.2) +
  theme_minimal() +
  labs(x = "-log10(p-value)", y = "Density") 
```

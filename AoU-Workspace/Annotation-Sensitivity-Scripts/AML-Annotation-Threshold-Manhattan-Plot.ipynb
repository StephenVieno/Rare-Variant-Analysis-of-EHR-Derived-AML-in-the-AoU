# AML ANNOTATION THRESHOLD MANHATTAN PLOT 
This script plots the results from the annotation sensitivity analysis to demonstrate which thresholds are optimal for each variant annotation category. In the AML results, the optimal thresholds were determined as follows:

- high-confidence loss-of-function variants from LOFTEE for the loss-of-function category;
- SpliceAI scores greater than 0.8 for splice variants;
- AlphaMissense scores greater than 0.546 for missense variants;
- and synonymous variants identified by VEP with SpliceAI scores less than 0.5 for the synonymous category.

```python
# Load necessary libraries
library(data.table)
library(ggplot2)
library(tidyverse)
```

```python
# Read in data files
lof <- fread("sensitivity_analysis/EUR_aml_phenotype_lof_chr21")
splice <- fread("sensitivity_analysis/EUR_aml_phenotype_splice_chr21")
missense <- fread("sensitivity_analysis/EUR_aml_phenotype_missense_chr21")
synonymous <- fread("sensitivity_analysis/EUR_aml_phenotype_synonymous_chr21")
```

```python
# Set plot dimensions
options(repr.plot.width = 20, 
        repr.plot.height = 10) 
```

## MANHATTAN PLOTS 

```python
# Plot loss-of-function results excluding Cauchy group
lof %>% 
  filter(Group != "Cauchy") %>% 
  mutate(
    neg_log_pvalue = -log10(Pvalue),
    max_MAF = factor(max_MAF)  
  ) %>% 
  ggplot(aes(x = Region, y = neg_log_pvalue, color = Group, shape = max_MAF)) +
  geom_jitter(size = 5, alpha = 0.5, width = 10) +
  theme_bw() +
  theme(
    legend.text = element_text(size = 14),
    legend.title = element_text(size = 18), 
    axis.marker
  ) + 
  paletteer::scale_colour_paletteer_d("LaCroixColoR::Lemon") 
```

```python
# Plot loss-of-function results excluding Cauchy group
lof %>% 
    filter(Group != "Cauchy") %>% 
    mutate(neg_log_pvalue = -log10(Pvalue)) %>% 
ggplot() +
geom_jitter(aes(x = Region, 
               y = neg_log_pvalue, 
               color = Group,
               size = 1, 
               alpha = 0.5), 
               width = 5) + 
theme_bw() + 
theme(legend.text = element_text(size = 14),
      legend.title = element_text(size = 18))
```

```python
# Plot splice results excluding Cauchy group
splice %>% 
    filter(Group != "Cauchy") %>% 
    mutate(neg_log_pvalue = -log10(Pvalue)) %>% 
ggplot() +
geom_jitter(aes(x = Region, 
               y = neg_log_pvalue, 
               color = Group, 
               size = neg_log_pvalue, 
               alpha = 0.5)) + 
theme_bw() + 
theme(legend.text = element_text(size = 14),
      legend.title = element_text(size = 18))
```

```python
# Plot splice results excluding Cauchy group
missense %>% 
    filter(Group != "Cauchy") %>% 
    mutate(neg_log_pvalue = -log10(Pvalue)) %>% 
ggplot() +
geom_jitter(aes(x = Region, 
               y = neg_log_pvalue, 
               color = Group, 
               size = 1, 
               alpha = 0.5), 
               width = 1) + 
theme_bw() + 
theme(legend.text = element_text(size = 14), 
      legend.title = element_text(size = 18))
```

```python
# Plot synonymous results excluding Cauchy group
synonymous %>% 
    filter(Group != "Cauchy") %>% 
    mutate(neg_log_pvalue = -log10(Pvalue)) %>% 
ggplot() +
geom_jitter(aes(x = Region, 
               y = neg_log_pvalue, 
               color = Group, 
               size = neg_log_pvalue, 
               alpha = 0.5)) + 
theme_minimal() + 
theme(legend.text = element_text(size = 14), 
      legend.title = element_text(size = 18))
```

```python
# Plot synonymous results excluding Cauchy group
synonymous %>% 
    filter(Group != "Cauchy") %>% 
    mutate(neg_log_pvalue = -log10(Pvalue)) %>% 
ggplot(aes(x = neg_log_pvalue, fill = Group, group = Group)) +
  geom_density(alpha = 0.2) +
  theme_minimal() +
  labs(x = "-log10(p-value)", y = "Density") 
```

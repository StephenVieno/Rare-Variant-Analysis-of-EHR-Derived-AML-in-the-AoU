```python
%%writefile annotate.R
#!/usr/bin/env Rscript

##########################################################################
# INPUT
##########################################################################

cat("========== START ==========\n")
args <- commandArgs(trailingOnly = TRUE)

file_DBsplit <- args[1]
DB_path <- args[2]
input_dir <- args[3]
output_dir <- args[4]
chr <- as.integer(args[5])

cat("CHROMOSOME:", chr, "\n")
cat("INPUT:", input_dir, "\n")
cat("DATABASE SPLIT INFORMATION:", file_DBsplit, "\n")
cat("FAVOR DATABASE:", DB_path, "\n")
cat("OUTPUT:", output_dir, "\n")
flush.console()

cat("\n========== PRE-EXTRACTION ==========\n")
cat("Database exists:", file.exists(DB_path), "\n")
cat("Split exists:", file.exists(file_DBsplit), "\n")
cat("Input directory exists:", dir.exists(input_dir), "\n")
cat("Output directory exists:", dir.exists(output_dir), "\n")
cat("Database size:", file.size(DB_path), "\n")
flush.console()

system("mkdir -p FAVOR", intern=TRUE)
system(paste("tar -xzf", DB_path, "-C FAVOR"), intern=TRUE)

cat("\n========== EXTRACTING FAVOR ==========\n")
dir.create("FAVOR", showWarnings = FALSE)

tar_status <- system2(
    "tar",
    c("-xzf", shQuote(DB_path), "-C", "FAVOR")
)

cat("tar exit status:", tar_status, "\n")

cat("\nFAVOR contents:\n")
print(list.files("FAVOR", recursive = TRUE, full.names = TRUE))
flush.console()

###########################################################################
# MAIN FUNCTION
###########################################################################

### anno channel (subset)
anno_colnum <- c(1,8:12,15,16,19,23,25:36)

### chromosome number
## annotate (seperate)
DB_info <- read.csv(file_DBsplit,header=TRUE)
chr_splitnum <- sum(DB_info$Chr==chr)

cat("\n========== CHECK XSV ==========\n")

xsv_path <- Sys.which("xsv")

cat("xsv path:", xsv_path, "\n")

if (xsv_path == "") {
    stop("xsv is NOT installed or not available in PATH")
}

cat("xsv is installed\n")

xsv_version <- system2(
    xsv_path,
    "--version",
    stdout = TRUE,
    stderr = TRUE
)

print(xsv_version)

for(kk in 1:chr_splitnum)
{

    cat("\n========== BLOCK", kk, "==========\n")
    
    input_file <- paste0(
        input_dir, "/variant_info_chr", chr, "_block", kk, ".csv"
    )
    
    favor_file <- paste0(
        "FAVOR/n/holystore01/LABS/xlin/Lab/xihao_zilin/FAVORDB/chr",
        chr, "_", kk, ".csv"
    )
    
    output_file <- paste0(
        output_dir, "/anno_chr", chr, "_block", kk, ".csv"
    )
    
    cat("INPUT:  ", input_file, "\n")
    cat("FAVOR:  ", favor_file, "\n")
    cat("OUTPUT: ", output_file, "\n")
    
    cat("Input exists: ", file.exists(input_file), "\n")
    cat("FAVOR exists: ", file.exists(favor_file), "\n")
    cat("xsv path:    ", Sys.which("xsv"), "\n")

    join_command <- paste(
    shQuote(xsv_path),
    "join --left",
    "VarInfo",
    shQuote(input_file),
    "variant_vcf",
    shQuote(favor_file),
    ">",
    shQuote(output_file),
    "2>&1"
    )
    
    cat("\n========== XSV JOIN ==========\n")
    cat("COMMAND:\n")
    cat(join_command, "\n")
    flush.console()
    
    join_output <- system(
        join_command,
        intern = TRUE
    )
    
    cat("XSV OUTPUT:\n")
    print(join_output)
    
    join_status <- attr(join_output, "status")
    
    if (is.null(join_status)) {
        join_status <- 0
    }
    
    cat("XSV JOIN EXIT STATUS:", join_status, "\n")
    flush.console()

}

## merge info
Anno <- paste0(output_dir,"/anno_chr",chr,"_block",seq(1:chr_splitnum),".csv ")
merge_command <- paste0("xsv cat rows ",Anno[1])

for(kk in 2:chr_splitnum)
{
	merge_command <- paste0(merge_command,Anno[kk])
}

merge_command <- paste0(merge_command,"> ",output_dir,"/anno_chr",chr,".csv")

system(merge_command)

## subset
anno_colnum_xsv <- c()
for(kk in 1:(length(anno_colnum)-1))
{
	anno_colnum_xsv <- paste0(anno_colnum_xsv,anno_colnum[kk],",")
}
anno_colnum_xsv <- paste0(anno_colnum_xsv,anno_colnum[length(anno_colnum)])

system(paste0("xsv select ",anno_colnum_xsv," ",output_dir,"/anno_chr",chr,".csv > ",output_dir,"/anno_chr",chr,"_STAARpipeline.csv"))
```

# SUBMIT COMMAND 

```python
import subprocess
import pandas as pd

# Create dataframe of size of exome plink files in GB 
chromosome = []
sizes = [] 

# Loop for all chromosome 
# for chrom in range(21, 22):
for chrom in [2, 4, 11, 15, 17, 19, 21]:
    
    # Extract size of genomic plink bed file in GB for each chromosome 
    command = [
        "gcloud", "storage", "ls",
        "-l",
        f"{os.environ['WORKSPACE_BUCKET']}/data/staar/gds/EUR_exome_chr{chrom}.gds",
        f"--billing-project={os.environ['GOOGLE_PROJECT']}"
    ]
    
    result = subprocess.run(command, capture_output=True, text=True)

    # Extract genomic size from string 
    lines = result.stdout.splitlines()
    line = lines[0].split()
    
    # Create list genomic size 
    sizes.append(line[0])
    chromosome.append(chrom)
    
# Deposit genomic file size into data frame 
genomic_size = pd.DataFrame({
    "chromosome": chromosome,
    "size": sizes
})

# Format genomic file size 
genomic_size['chromosome'] = genomic_size['chromosome'].astype('int')
genomic_size['size'] = (genomic_size['size'].astype('int') / 1e9) 

print(genomic_size)
```

```python
import shlex

# Logging directory
LOGGING = f"{os.environ['WORKSPACE_BUCKET']}/data/logging"

# Dsub script
SCRIPT = "annotate.R"
PATH = f"{os.environ['WORKSPACE_BUCKET']}/data/scripts/{SCRIPT}"

# Paths
SPLIT = f"{os.environ['WORKSPACE_BUCKET']}/data/staar/axillary/FAVORdatabase_chrsplit.csv"

MAX_DISK_SIZE = 100
EXTRA_DISK_SIZE = 10

# Loop to submit dsub command
# for CHROM in range(21, 22): for CHROM in [2, 4, 11, 15, 17, 19]:
for CHROM in [21]:

    print(f"Submitting dsub command for chromosome {CHROM}")

    DISK_SIZE = (
        round(
            float(genomic_size.loc[genomic_size["chromosome"] == CHROM, "size"
                ].iloc[0]
            )
        )
        + EXTRA_DISK_SIZE
    )
    DISK_SIZE = 80
    print(f"Requesting disk size: {DISK_SIZE}")

    if DISK_SIZE > MAX_DISK_SIZE:
        print(f"Disk too large for chromosome {CHROM}")
        break

    if DISK_SIZE > 1000:
        print(f"Disk too large for chromosome {CHROM}")
        break

    # File paths
    DATABASE = f"{os.environ['WORKSPACE_BUCKET']}/data/staar/favor/FAVOR_essential_chr{CHROM}.tar.gz"
    INPUT = f"{os.environ['WORKSPACE_BUCKET']}/data/staar/variant_info/chr{CHROM}"
    OUTPUT = f"{os.environ['WORKSPACE_BUCKET']}/data/staar/annotation/chr{CHROM}"

    # R script command
    R_COMMAND = 'Rscript "$SCRIPT" "$SPLIT" "$DATABASE" "$INPUT" "$OUTPUT" "$CHROM"'
    print(R_COMMAND)
    command = [
        "aou_dsub",
            "--image", f"{os.environ['ARTIFACT_REGISTRY_DOCKER_REPO']}/vieno99/favor:v1.0",
            "--boot-disk-size", "5",
            "--disk-size", str(DISK_SIZE),
            "--machine-type", "n1-standard-1",
            "--logging", LOGGING,
            "--input", f"SCRIPT={PATH}",
            "--input", f"SPLIT={SPLIT}",
            "--input", f"DATABASE={DATABASE}",
            "--input-recursive", f"INPUT={INPUT}",
            "--output-recursive", f"OUTPUT={OUTPUT}",
            "--env", f"CHROM={CHROM}",
            "--command", R_COMMAND,
    ]

    # Run in same shell as source command
    dsub_command = subprocess.run(
        f"source ~/aou_dsub.bash && {shlex.join(command)}",
        shell=True,
        executable="/bin/bash",
        capture_output=True,
        text=True
    )

    print(dsub_command.stdout)
    print(dsub_command.stderr)
```

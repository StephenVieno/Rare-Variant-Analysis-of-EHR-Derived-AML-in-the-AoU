```r
# Load packages 
library(data.table)
library(tidyverse)
library(readr)
library(ggplot2)
```

```r
# Ensure a clean directory for the logistic regression loop 
system("rm -rf logistic_summary")
system("mkdir -p logistic_summary")
# Copy files from workspace bucket 
system("gsutil cp gs://fc-secure-bf507049-90b6-43f8-86ea-c3ad9e823ffe/data/gene_burden_analysis/burden_logistic_summary/* logistic_summary")
```

```r
# List of file present in the logistic summary folder 
files <- list.files(path="logistic_summary")

###############################################################################################################
# Combine and format rare variant polygenic risk score results 
###############################################################################################################
data <- list() 
for (file in files) {
  print(paste0("Processing ", file, " ...."))
  summary <- fread(paste0("logistic_summary/", file)) 
  summary <- summary[2,]
  name <- gsub(".tsv$", "", file)  
  summary$Summary <- as.vector(name)
  assign(name, summary)          
  flush.console()                
  Sys.sleep(0.5)                  
  data[[name]] <- summary          
}

data <- do.call(rbind, data)
data <- janitor::clean_names(data)
data$bonferroni <- p.adjust(data$pr_z, method = "bonferroni")
data$summary <- gsub("-logistic-summary", "", data$summary)
data$summary <- gsub("-", " ", data$summary)
data <- arrange(select(data, summary, estimate, std_error, pr_z, bonferroni), pr_z)
colnames(data) <- c('Gene-Set Description', 'Log Odds', 'Standard Error', 'P-value', 'Bonferroni P-value')
write_tsv(data, "exome_burden_logistic_summary.tsv")
```

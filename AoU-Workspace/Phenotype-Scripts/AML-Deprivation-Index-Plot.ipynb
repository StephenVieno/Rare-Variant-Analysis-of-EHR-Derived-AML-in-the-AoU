```r
library(data.table)
library(tidyverse)
library(readr)
library(ggplot2)
library(forcats)
library(patchwork)
```

```r
data <- fread("eur_aml_phenotype_deprivation_index.txt")
```

```r
options(repr.plot.width = 20, repr.plot.height = 5)
```

```r
quantiles <- 
data %>% 
  group_by(aml_phenotype) %>% 
  summarise(
    q05 = quantile(deprivation_index, 0.05, na.rm = TRUE),
    q25 = quantile(deprivation_index, 0.25, na.rm = TRUE),
    q50 = quantile(deprivation_index, 0.50, na.rm = TRUE),
    q75 = quantile(deprivation_index, 0.75, na.rm = TRUE),
    q95 = quantile(deprivation_index, 0.95, na.rm = TRUE)
  ) 
```

```r
pal <- paletteer::paletteer_d("LaCroixColoR::Lemon")
```

```r
df <- data %>% 
  mutate(
    aml_phenotype = as.factor(aml_phenotype),
    status = ifelse(aml_phenotype == 1, "Case", "Control")
  )

plot1 <- 
ggplot(df) + 
  geom_density(aes(
    x = deprivation_index, 
    fill = status,
    alpha = 0.5
  )) + 
  
  # First set of quantile lines (solid)
  geom_vline(
    xintercept = as.numeric(quantiles[1, 3]),
    color = "#172869FF", size = 2, alpha = 0.5
  ) +
  geom_vline(
    xintercept = as.numeric(quantiles[1, 4]),
    color = "#172869FF", size = 2, alpha = 0.7
  ) +
  geom_vline(
    xintercept = as.numeric(quantiles[1, 5]),
    color = "#172869FF", size = 2, alpha = 0.5
  ) +

  # Second set of quantile lines with different alphas
  geom_vline(
    xintercept = as.numeric(quantiles[2, 3]),
    color = "#1BB6AFFF", size = 2, alpha = 0.5
  ) +
  geom_vline(
    xintercept = as.numeric(quantiles[2, 4]),
    color = "#1BB6AFFF", size = 2, alpha = 0.7
  ) +
  geom_vline(
    xintercept = as.numeric(quantiles[2, 5]),
    color = "#1BB6AFFF", size = 2, alpha = 0.5
  ) +

  theme_bw() + 
  scale_fill_manual(values = c(pal[4], pal[6])) + 
  labs(x = "Deprivation Index", y = "Density", fill = "Status") +
  guides(alpha = "none")
```
